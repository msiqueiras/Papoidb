## Inserção e Ordenação

```python
print("Inserindo 20, 10, 40, 30...")
arvore.insert(20, 200)
arvore.insert(10, 100)
arvore.insert(40, 400)
arvore.insert(30, 300)

# Verifica a ordenação na folha
raiz = arvore.root
print("\nChaves na folha (devem estar ordenadas):")
for i in range(raiz.getNumKeys()):
    print(f"Índice {i}: Chave {raiz.getKey(i)} -> Ponteiro {raiz.getPointer(i)}")
```
